```python
from ultralytics import YOLO
import os, torch, cv2

weights_path_nano = "yolov8n.pt"  
weights_path_small = "yolov8s.pt"
weights_path_large = "yolov8l.pt"
try:
    if os.path.exists(f"C:/Users/there/OneDrive/Desktop/Object Detection/Detection Models/runs"):
        os.remove(f"C:/Users/there/OneDrive/Desktop/Object Detection/Detection Models/runs")
except OSError:
    pass


device = "cuda" if torch.cuda.is_available() else "cpu"

user_input = input("-- Enter your Preferrence of Model Weights ( Nano / Large / Small ) : ").strip().title()
try:
    if user_input == "Nano":
        weights = weights_path_nano
    elif user_input == "Large":
        weights = weights_path_large
    elif user_input == "Small":
        weights = weights_path_small    
    else:
        print(f"Invalid Input")
except BaseException as e:
    print(f"Invalid Input")


path1 = os.path.join("..","Images", "Children.jpg")
path2 = os.path.join("..","Images", "Pets.png")
path3 = os.path.join("..","Images", "Tools.jpg")

model = YOLO(weights)

image1 = model.predict(path1, show=True, save=True) 
image2 = model.predict(path2, show=True, save=True) 
image3 = model.predict(path3, show=True, save=True) 

cv2.waitKey(0)
```
